```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
data = {
    'City': ['New York', 'Los Angeles', 'Chicago', 'Houston', 'Phoenix',
             'Philadelphia', 'San Antonio', 'San Diego', 'Dallas', 'San Jose'],
    'Latitude': [40.7128, 34.0522, 41.8781, 29.7604, 33.4484,
                 39.9526, 29.4241, 32.7157, 32.7767, 37.3382],
    'Longitude': [-74.0060, -118.2437, -87.6298, -95.3698, -112.0740,
                  -75.1652, -98.4936, -117.1611, -96.7970, -121.8863]
}

df = pd.DataFrame(data)
print(df)
```

```python
X=df[['Latitude','Longitude']]

kmeans= KMeans(n_clusters=3, random_state=42)
df['Clusters']=kmeans.fit_predict(X)
```

```python
centroids=kmeans.cluster_centers_
```

```python
plt.figure(figsize=(8,6))

plt.scatter(df['Longitude'],df['Latitude'],c=df['Clusters'],cmap='viridis',s=100)

plt.scatter(centroids[:,1],centroids[:,0],marker='X',s=200,color='red',label='Centroids')

for i,city in enumerate(df['City']):
       plt.text(df['Longitude'][i] + 0.3, df['Latitude'][i] + 0.3, city, fontsize=8)

plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.title("K-Means Clustering of Cities")
plt.legend()
plt.grid()
plt.show()
```
