```python
#• Use Iris dataset with 150 samples and 4 features (use only the first two features: 
#Sepal Length and Sepal Width). 
#• K-means clusters the data into 3 clusters, corresponding to the 3 species in the Iris 
#dataset.
```

```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
```

```python
#Load Iris Dataset
iris=load_iris()
X=iris.data[:, :2]
```

```python
#Apply K Means with 3 clusters
kmeans=KMeans(n_clusters=3, random_state=42)
kmeans.fit(X)
```

```python
labels=kmeans.labels_
centroids=kmeans.cluster_centers_
```

```python
plt.scatter(X[:, 0], X[:, 1], c=labels,cmap='viridis')
```

```python
plt.scatter(centroids[:, 0],centroids[:, 1],c='red',marker='X',s=200)

plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.title("K-Means Clustering on Iris Dataset (k=3)")

plt.show()
```
