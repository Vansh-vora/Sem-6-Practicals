```python
#Write a machine learning code to create a scatter plot for the following data points:
# x = [2, 4, 6, 8, 10]
# y = [3, 6, 9, 12, 15]
```

```python
import matplotlib.pyplot as plt
```

```python
x=[2,4,6,8,10]
y=[3,6,9,12,15]
```

```python
plt.scatter(x,y)
plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.title("Scatter Plot")
plt.grid()
plt.show()
```
