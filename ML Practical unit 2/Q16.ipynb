```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
transaction_amounts = [10, 15, 20, 25, 30, 35, 1000, 2000]

# Box plot
plt.boxplot(transaction_amounts)
plt.title("Transaction Amounts Box Plot")
plt.show()

# Outlier detection
Q1 = np.percentile(transaction_amounts, 25)
Q3 = np.percentile(transaction_amounts, 75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = [x for x in transaction_amounts if x < lower or x > upper]
print("Outliers:", outliers)
```

```python
cv_scores_model_1 = [0.82, 0.84, 0.83, 0.81, 0.95]
cv_scores_model_2 = [0.88, 0.87, 0.89, 0.86, 0.88]

# Box plot
plt.boxplot([cv_scores_model_1, cv_scores_model_2], labels=['Model 1', 'Model 2'])
plt.title("CV Scores Comparison")
plt.show()

# Function to find outliers
def find_outliers(data):
    Q1 = np.percentile(data, 25)
    Q3 = np.percentile(data, 75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    return [x for x in data if x < lower or x > upper]

print("Model 1 Outliers:", find_outliers(cv_scores_model_1))
print("Model 2 Outliers:", find_outliers(cv_scores_model_2))
```

```python
daily_returns = [0.01, 0.02, 0.015, -0.03, 0.02, -0.1, 0.03, 0.08]

# Box plot
plt.boxplot(daily_returns)
plt.title("Daily Returns Box Plot")
plt.show()

# Outliers
print("Outliers:", find_outliers(daily_returns))
```
