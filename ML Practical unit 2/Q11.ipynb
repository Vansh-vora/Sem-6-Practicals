```python
#2. A house price dataset contains a column with property prices. 100 Prices should 
#be generated randomly.
#Display the histogram of it with the following conditions:
# Bins should be 10
# Give title “Price distribution
# Edge color should be gra
# Color should be pink
# Give x and y axis labels accordingly
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)
prices=np.random.randint(50000,500000,100)

df=pd.DataFrame({'Price':prices})

print(df.head())
```

```python
plt.hist(df['Price'],bins=10,color='pink',edgecolor='grey')
plt.title("Price Distribution")
plt.xlabel("Price")
plt.ylabel("Frequency")
plt.show()
```
