```python
import matplotlib.pyplot as plt
```

```python
data=[10,20,20,25,30,35,40,50]
```

```python
#Create Histogram
plt.hist(data,bins=5,edgecolor='black')
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.title("Histogram")
plt.show()
```
