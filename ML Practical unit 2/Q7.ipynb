```python
#Generate a bar chart for a dataset where the categories are ['Dog', 'Cat', 'Bird'] and 
#their respective counts are [3, 5, 2].
```

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data= ['Dog','Cat','Bird']
y=[3,5,2]
```

```python
plt.bar(data,y,edgecolor="red")
plt.xlabel("Animals")
plt.ylabel("Count")
plt.show()
```
