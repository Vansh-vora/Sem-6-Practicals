```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)

# Create dataset
data = {
    'Feature1': np.random.rand(100),
    'Feature2': np.random.rand(100) * 2,
    'Feature3': np.random.rand(100) * 3,
    'Target': np.random.rand(100) * 4
}

df = pd.DataFrame(data)

# Correlation matrix
corr_matrix = df.corr()

# Heatmap
plt.imshow(corr_matrix)

plt.xticks(range(len(corr_matrix.columns)), corr_matrix.columns)
plt.yticks(range(len(corr_matrix.columns)), corr_matrix.columns)

plt.colorbar()
plt.title("Heatmap of Feature Correlation")

plt.show()
```

```python
# Generate data
true_values = np.random.rand(10)
predicted_values = true_values + np.random.normal(0, 0.1, 10)

# Create DataFrame
df2 = pd.DataFrame({
    'True': true_values,
    'Predicted': predicted_values
})

# Correlation matrix
corr_matrix2 = df2.corr()

# Heatmap
plt.imshow(corr_matrix2)

plt.xticks(range(len(corr_matrix2.columns)), corr_matrix2.columns)
plt.yticks(range(len(corr_matrix2.columns)), corr_matrix2.columns)

plt.colorbar()
plt.title("Heatmap of True vs Predicted Values")

plt.show()
```
