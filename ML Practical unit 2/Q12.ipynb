```python
# {'Age': [25, 30, 35, None, 28, 32, 45, None, 30, 28]} Create a histogram after 
#deleting or impute missing values.
```

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data={
    'Age':[25,30,35,None,28,32,45,None,30,28]
}
df=pd.DataFrame(data)
print(df)
```

```python
df_clean=df.dropna()
```

```python
plt.hist(df_clean['Age'],bins=5,edgecolor='black')
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
df_imputed=df.copy()
mean_age=df_imputed['Age'].mean()
df_imputed['Age'].fillna(mean_age)

print("Mean Age",mean_age)
plt.hist(df_imputed['Age'], bins=5, edgecolor='black')

plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Age Distribution (After Imputing Missing Values)")

plt.show()
```
