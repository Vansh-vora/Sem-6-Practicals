```python
#Write a machine learning code to visualize the frequency of letters in the dataset 
#{'Letters': ['A', 'B', 'A', 'C', 'B', 'C', 'A']}.
```

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data={
    'Letters':['A','B','A','C','B','C','A']}
df=pd.DataFrame(data)
print(df)
```

```python
frequency=df['Letters'].value_counts()
```

```python
plt.bar(frequency.index,frequency.values,edgecolor='black')
plt.xlabel("Letters")
plt.ylabel("Frequency")
plt.title("Frequency of Letters")
plt.show()
```
