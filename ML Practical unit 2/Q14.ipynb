```python
#Create a group bar graph and Stacked bar graph for the following data:
#Shape = ['Square1', 'Rect', 'Square2']
#Length = [10, 15, 20]
#Height = [12, 18, 25]
```

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
shape=['Square1','Rect','Sqaure2']
length=[10,15,20]
height=[12,18,25]
```

```python
x=np.arange(len(shape))
width=0.35
```

```python
plt.bar(x - width/2, length, width, label='Length')
plt.bar(x + width/2, height, width, label='Height')

plt.xlabel("Shapes")
plt.ylabel("Values")
plt.title("Grouped Bar Graph")

plt.xticks(x, shape)
plt.legend()

plt.show()
```
