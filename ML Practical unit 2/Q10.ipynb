```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data ={
    'Age':[22,35,34,56,67,54,43,56,78,10]
           }
df=pd.DataFrame(data)
print(df)
```

```python
plt.hist(df['Age'],bins=5,color='blue',edgecolor='black')
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("No of Customers")
plt.show()
```
